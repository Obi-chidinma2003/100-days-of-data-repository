**IMPORTING THE LIBRARIES**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

EXPLANATION OF THIS ANALYSIS
Univariate analysis involves working with one column or content of the dataset. I.e one content visualisation per code.

**IMPORTING THE TIPS DATASET**

```python
df=sns.load_dataset("tips")
df.head()
```

**COUNT PLOT FOR CATEGORICAL VARIABLES**

```python
sns.countplot(df['sex'])
```

**PIE CHART FOR CATEGORICAL VARIABLES**

```python
df['sex'].value_counts().plot(kind='pie',autopct='%2F')
#autopct=%2F shows the percent of the pie for each value.
```

```python
df['sex'].value_counts().plot(kind='hist')
```

```python
plt.hist(df['total_bill'])
```

```python
sns.kdeplot(df['total_bill'])
```

```python
sns.scatterplot(df['total_bill'])
```
