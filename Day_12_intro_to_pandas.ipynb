**IMPORTING THE REQUIRED LIBRARIES**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**DISPLAYING THE FIRST 10 CONTENTS OF HE DATASET**

```python
path=("/content/drive/MyDrive/Dataset (1)/Titanic-Dataset.csv")
df=pd.read_csv(path)
df.head(10)
```

```python
df.describe()
```

**REMOVING SOME IRRELEVANT COLUMNS FROM THE DATASET**

```python
df.drop(['Cabin','SibSp', 'Ticket'],axis=1).head()
```

**CHECKING FOR MISSING VALUES**

```python
df.isnull()
```

**CHECKING THE SUM OF MISSING VALUES**

```python
df.isnull().sum()
```

**DISPLAYING THE NUMBER OF PEOPLE WHO EMBARKED ON THE JOURNEY USING BAR CHART**

```python
pd.value_counts(df['Embarked']).plot.bar()
```

**DISPLAYING THE NUMBER OF PEOPLE WHO SURVIVED THE TITANIC USING BAR CHART**

```python
pd.value_counts(df['Survived']).plot.bar()
```

**AVERAGE NUMBER OF PEOPLE WHO SURVIVED THE TITANIC**

```python
df['Survived'].mean()
```

**GROUPING BASED ON GENDER FOR EACH COLUMN**

```python
df.groupby(['Sex']).mean(numeric_only=True)
```

**GROUPING BASED ON SEX AND THE PLANE CLASS(1ST CLASS TO 3RD CLASS)**

```python
df.groupby(['Sex','Pclass']).mean(numeric_only=True)
```

**NUMBER OF MALE AND FEMALE WHO ARE LESS THAN 18 AND SURVIVED THE TITANIC.**

```python
df[df['Age']<18].groupby(['Sex','Pclass']).mean(numeric_only=True)
```

**NO OF MALE AND FEMALE LESS THAN I8 WHO SURVIVED THE TITANIC VISUALISED ON A BAR CHART.**



```python
df[df['Age']<18].groupby(['Sex','Pclass']).mean(numeric_only=True).plot.bar()
```
