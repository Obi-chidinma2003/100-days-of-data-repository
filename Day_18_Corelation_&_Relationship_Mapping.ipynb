**IMPORTING LIBRARIES**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

**READING THE DATASET**

```python
Path=("/content/drive/MyDrive/Dataset/StudentsPerformance.csv")
df=pd.read_csv(Path)
df.head()
```

**DISPLAYING THE TITLE OF EACH COLUMNS**

```python
df.columns
```

**HIGLIGHTING/SEPERATING SOME COLUMNS TO FORM A NEW TABLE**

```python
df=df[['math score', 'reading score','writing score']]
df.head()
```

**CHECKING FOR CORRELATIONS**

```python
cor =df.corr()
cor
```

**SHOWING CORRELATION USING HEATMAP**

```python
sns.heatmap(data=cor,annot=True,)
```
